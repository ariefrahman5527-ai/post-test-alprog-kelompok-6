```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Membaca data dari file CSV
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# Mengonversi kolom ke tipe numerik dan membersihkan baris kosong (NaN)
df['Rank'] = pd.to_numeric(df['Rank'], errors='coerce')
df['M_secondary_educ'] = pd.to_numeric(df['M_secondary_educ'], errors='coerce')
df_clean = df.dropna(subset=['Rank', 'M_secondary_educ'])

# Menentukan variabel sumbu x dan y
x = df_clean['Rank']
y = df_clean['M_secondary_educ']

# Membuat scatter plot untuk melihat persebaran data
plt.scatter(x, y, color='steelblue', alpha=0.7, label='Sebaran Data Negara')

# Menghitung koefisien regresi linear dengan numpy untuk membuat garis tren korelasi
m, b = np.polyfit(x, y, 1)
plt.plot(x, m*x + b, color='red', label='Garis Tren Korelasi')

# Menambahkan judul, label sumbu, dan legenda
plt.title('Korelasi Rank GII dengan Pendidikan Menengah Laki-laki')
plt.xlabel('Rank GII')
plt.ylabel('Pendidikan Menengah Laki-laki (%)')
plt.legend()

# Merapikan tata letak dan menampilkan grafik
plt.tight_layout()
plt.show()
```
