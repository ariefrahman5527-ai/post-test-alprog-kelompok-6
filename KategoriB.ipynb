```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Membaca dataset dari file CSV
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# Memastikan kolom bernilai numerik agar bisa dibandingkan
df['Adolescent_birth_rate'] = pd.to_numeric(df['Adolescent_birth_rate'], errors='coerce')
df['Maternal_mortality'] = pd.to_numeric(df['Maternal_mortality'], errors='coerce')

# Menyaring data berdasarkan dua kondisi yang diminta
filtered = df[(df['Adolescent_birth_rate'] < 10) & (df['Maternal_mortality'] > 50)]

# Menentukan indeks koordinat x dan lebar bar
x = np.arange(len(filtered['Country']))
width = 0.35

# Membentuk grouped bar chart
plt.bar(x - width/2, filtered['Adolescent_birth_rate'], width, label='Adolescent Birth Rate', color='steelblue')
plt.bar(x + width/2, filtered['Maternal_mortality'], width, label='Maternal Mortality', color='salmon')

# Mengatur teks sumbu, judul, dan legenda grafik
plt.title('Adolescent Birth Rate vs Maternal Mortality')
plt.xlabel('Negara')
plt.ylabel('Nilai')
plt.xticks(x, filtered['Country'])
plt.legend()

# Mengoptimalkan tata letak agar rapi
plt.tight_layout()

# Menampilkan hasil visualisasi
plt.show()
```
