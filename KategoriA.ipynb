```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Membaca data dari file CSV
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# Mengelompokkan data dan menghitung nilai minimum dengan fungsi agregasi numpy
min_data = df.groupby('Human_development')['F_Labour_force'].agg(np.min)

# Membuat bar chart dari data hasil agregasi
plt.bar(min_data.index, min_data.values, color='steelblue', edgecolor='black')

# Menambahkan judul dan label pada grafik
plt.title('Nilai Minimum F_Labour_force per Kelompok Human Development')
plt.xlabel('Kelompok Human Development')
plt.ylabel('F_Labour_force (Minimum)')

# Mengatur rotasi teks sumbu x agar tidak bertumpuk
plt.xticks(rotation=45)
plt.tight_layout()

# Menampilkan hasil visualisasi
plt.show()
```
