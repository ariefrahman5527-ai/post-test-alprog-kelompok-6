```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Membaca data dari file CSV
df = pd.read_csv('Kelas A_Gender Inequality Index.csv')

# Mengonversi kolom target ke tipe numerik dan menangani nilai kosong
df['F_Labour_force'] = pd.to_numeric(df['F_Labour_force'], errors='coerce')

# Membuat boxplot berdasarkan pengelompokan Human_development
df.boxplot(column='F_Labour_force', by='Human_development', grid=False, patch_artist=True)

# Menghapus sub-judul bawaan dari pandas agar tidak tumpang tindih
plt.suptitle('')

# Menetapkan judul utama dan label sumbu pada grafik
plt.title('Sebaran Partisipasi Angkatan Kerja Perempuan')
plt.xlabel('Kelompok Human Development')
plt.ylabel('F_Labour_force (%)')

# Mengoptimalkan tata letak visual
plt.tight_layout()

# Menampilkan hasil visualisasi
plt.show()
```
